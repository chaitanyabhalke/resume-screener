```python
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
def preprocess(text):
    # Step 1 - Lowercase
    text = text.lower()
    
    # Step 2 - Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    
    # Step 3 - Tokenize
    tokens = word_tokenize(text)
    
    # Step 4 - Remove stopwords
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]
    
    # Step 5 - Lemmatize
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    
    return ' '.join(tokens)
```

```python
sample = "I have been Working with Python, PYTHON and python since 2022!!! I am experienced in Machine Learning."

print("Before:", sample)
print("After:", preprocess(sample))
```

```python
resume = """
I am a Python developer with 2 years of experience in machine learning and data analysis.
I have worked with scikit-learn, pandas, and numpy for building predictive models.
I am familiar with natural language processing and have built text classification projects.
I have experience with SQL databases and data visualization using matplotlib and seaborn.
I am a good communicator and team player with strong problem solving skills.
"""

job_description = """
We are looking for a Python developer with experience in machine learning and deep learning.
The candidate should have strong knowledge of NLP and text processing.
Experience with TensorFlow or PyTorch is required.
The candidate should be familiar with SQL and data visualization tools.
Strong communication and teamwork skills are essential.
Experience with cloud platforms like AWS or Azure is a plus.
"""

print("Resume processed:")
print(preprocess(resume))
print()
print("JD processed:")
print(preprocess(job_description))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

processed_resume = preprocess(resume)
processed_jd = preprocess(job_description)

tfidf_matrix = vectorizer.fit_transform([processed_resume, processed_jd])

print("Vocabulary size:", len(vectorizer.vocabulary_))
print("Matrix shape:", tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

score = cosine_similarity(tfidf_matrix[0], tfidf_matrix[1])

print("Match Score:", round(score[0][0] * 100, 2), "%")
```

```python
# ============================================
# AI POWERED RESUME SCREENER
# ============================================

def screen_resume(resume_text, jd_text):
    
    # Stage 1 - Preprocess both texts
    processed_resume = preprocess(resume_text)
    processed_resume = processed_resume.replace('mysql', 'sql').replace('postgresql', 'sql')
    processed_jd = preprocess(jd_text)
    
    # Stage 2 - TF-IDF Vectorization
    vectorizer = TfidfVectorizer()
    tfidf_matrix = vectorizer.fit_transform([processed_resume, processed_jd])
    
    # Stage 3 - Cosine Similarity
    score = cosine_similarity(tfidf_matrix[0], tfidf_matrix[1])
    match_score = round(score[0][0] * 100, 2)
    
    # Stage 4 - Keyword Gap Analysis
    jd_words = set(processed_jd.split())
    resume_words = set(processed_resume.split())
    junk_words = {'looking', 'candidate', 'essential', 'required', 'plus',
              'like', 'tool', 'platform', 'knowledge', 'communication',
              'mandatory', 'excellent', 'must', 'strong', 'know', 'hand',
              'good', 'also', 'well', 'need', 'role', 'using', 'use',
              'work', 'working', 'year', 'experience', 'skill', 'ability','developer', 'text', 'processing', 'power', 'bi', 'visualization'}
    missing_keywords = jd_words - resume_words - junk_words
    
    # Stage 5 - LLM Feedback
    prompt = f"""
    You are an expert HR consultant and resume coach.

    Here is a candidate's resume:
    {resume_text}

    Here is the job description:
    {jd_text}

    The resume match score is: {match_score}%
    The missing keywords are: {', '.join(missing_keywords)}

    Give the candidate a short, honest, actionable feedback in 4-5 lines.
    Tell them what they did well, what is missing, and what they should add.
    """
    
    response = model.generate_content(prompt)
    
    # Final Output
    print("=" * 50)
    print(f"MATCH SCORE: {match_score}%")
    print("=" * 50)
    print("\nMISSING KEYWORDS:")
    for word in missing_keywords:
        print(f"  - {word}")
    print("\nAI FEEDBACK:")
    print(response.text)
    print("=" * 50)

# Run it
screen_resume(resume, job_description)
```
